```python
graph = {
    'S': ['D', 'A'],
    'D': ['E'],
    'E': ['G'],
    'A': ['C'],
    'C': ['G'],
    'G': []
}

def dfs(node, goal, visited, path):
    visited.add(node)
    path.append(node)
    if node == goal:
        return path[:]
    for neighbor in graph.get(node, []):
        if neighbor not in visited:
            result = dfs(neighbor, goal, visited, path)
            if result:
                return result
    path.pop()
    return None

path = dfs('S', 'G', set(), [])
print(" -> ".join(path))
```

```python
from collections import deque

graph = {
    'S': ['A', 'D'],
    'A': ['C'],
    'C': ['G'],
    'D': ['E'],
    'E': ['G'],
    'G': []
}

def bfs(start, goal):
    visited = {start}
    queue = deque([[start]])
    while queue:
        path = queue.popleft()
        node = path[-1]
        if node == goal:
            return path
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(path + [neighbor])
    return None

path = bfs('S', 'G')
print(path, len(path) - 1)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

rng = np.random.default_rng(42)
X = rng.uniform(-3, 3, size=(30,1))
y_true = np.sin(X)
y = y_true + rng.normal(0, 0.3, size=(30,1))

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.35, random_state=7)

for deg in [1, 15]:
    poly = PolynomialFeatures(degree=deg, include_bias=False)
    X_train_p = poly.fit_transform(X_train)
    X_test_p = poly.transform(X_test)
    model = LinearRegression()
    model.fit(X_train_p, y_train)
    train_mse = mean_squared_error(y_train, model.predict(X_train_p))
    test_mse = mean_squared_error(y_test, model.predict(X_test_p))
    print(f'Degree {deg}: train MSE={train_mse:.4f}, test MSE={test_mse:.4f}')
```

```python
facts = {'rain'}
rules = [
    (('rain',), 'wet_ground'),
    (('wet_ground',), 'slippery'),
    (('slippery',), 'slow_down'),
]

changed = True
while changed:
    changed = False
    for conditions, conclusion in rules:
        if all(c in facts for c in conditions) and conclusion not in facts:
            facts.add(conclusion)
            changed = True

print(facts, len(facts))
```
